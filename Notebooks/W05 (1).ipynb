````python
from pathlib import Path
import duckdb

PROJECT_ROOT = Path(".").resolve()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
DB_PATH = PROJECT_ROOT / "data" / "exoplanets.duckdb"
ART_DIR = PROJECT_ROOT / "artifacts"
DOCS_DIR = PROJECT_ROOT / "docs"

RAW_DIR.mkdir(parents=True, exist_ok=True)
ART_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

con = duckdb.connect(str(DB_PATH))

raw_csv = RAW_DIR / "pscomppars.csv"
if not raw_csv.exists():
    raise FileNotFoundError(
        f"No encuentro {raw_csv}. Necesitas el CSV de W01/W02."
    )

def sql_quote(s: str) -> str:
    return "'" + s.replace("'", "''") + "'"

con.execute(f'''
CREATE OR REPLACE VIEW raw_ps AS
SELECT * FROM read_csv_auto({sql_quote(str(raw_csv.resolve()))})
''')

# Si ya tienes Silver/Facts de W03B, perfecto. Si no, los construimos rápido aquí (idéntico a W03B).
con.execute("DROP TABLE IF EXISTS silver_planet")
con.execute('''
CREATE TABLE silver_planet AS
SELECT
  pl_name,
  hostname,
  discoverymethod,
  disc_year,
  sy_snum,
  sy_pnum,
  sy_dist,
  ra,
  dec,
  pl_orbper,
  pl_rade,
  pl_bmasse,
  pl_eqt,
  st_teff,
  st_rad,
  st_mass
FROM raw_ps
WHERE pl_name IS NOT NULL
  AND hostname IS NOT NULL
  AND (disc_year IS NULL OR (disc_year BETWEEN 1980 AND 2026))
  AND (pl_rade  IS NULL OR (pl_rade  > 0 AND pl_rade  <= 30))
  AND (pl_bmasse IS NULL OR (pl_bmasse > 0))
''')

con.execute("DROP TABLE IF EXISTS dim_host_full")
con.execute('''
CREATE TABLE dim_host_full AS
SELECT
  hostname,
  MAX(sy_dist)  AS sy_dist,
  MAX(ra)       AS ra,
  MAX(dec)      AS dec,
  MAX(st_teff)  AS st_teff,
  MAX(st_rad)   AS st_rad,
  MAX(st_mass)  AS st_mass
FROM silver_planet
GROUP BY hostname
''')

con.execute("DROP TABLE IF EXISTS fact_planet")
con.execute('''
CREATE TABLE fact_planet AS
SELECT DISTINCT
  pl_name,
  hostname,
  discoverymethod,
  disc_year,
  pl_orbper,
  pl_rade,
  pl_bmasse,
  pl_eqt
FROM silver_planet
''')

con.sql("SELECT COUNT(*) AS n_fact, COUNT(DISTINCT pl_name) AS n_pl, COUNT(DISTINCT hostname) AS n_hosts FROM fact_planet").show()
````

## 1) Qué vamos a mirar en un plan (lectura rápida)

Cuando mires `EXPLAIN`, busca:
- **SCAN**: ¿está leyendo demasiadas filas/columnas?
- **FILTER**: ¿se aplica temprano o tarde?
- **JOIN**: ¿qué tipo? (HASH JOIN típico en analítica)
- **GROUP BY**: ¿cuántas filas entran?
- **CARDINALIDAD**: cuántas filas pasan por cada operador

`EXPLAIN` = estimado  
`EXPLAIN ANALYZE` = real (ejecuta y muestra cardinalidad real + tiempos)

````python
# TU TURNO 1: pega aquí una consulta tipo "métrica" y mira su plan.
# Debe tener: WHERE + GROUP BY (como en W02A).
# Ejemplo de idea: por disc_year o por discoverymethod.

q = '''
SELECT discoverymethod, COUNT(*) AS n_planets
FROM fact_planet
WHERE disc_year >= 2010
GROUP BY discoverymethod
ORDER BY n_planets DESC
LIMIT 5
'''

con.sql("EXPLAIN " + q).show()
# Si corre rápido:
con.sql("EXPLAIN ANALYZE " + q).show()
````

````python
# TU TURNO 2: escribe dos consultas equivalentes:
# A) con SELECT * y B) con solo 3 columnas.
# Luego compara sus planes.

qA = '''
SELECT * FROM fact_planet WHERE disc_year >= 2010 LIMIT 100
'''
qB = '''
SELECT pl_name, discoverymethod, disc_year FROM fact_planet WHERE disc_year >= 2010 LIMIT 100
'''

con.sql("EXPLAIN " + qA).show()
con.sql("EXPLAIN " + qB).show()
````

````python
from pathlib import Path
import duckdb
import json
from datetime import datetime

PROJECT_ROOT = Path(".").resolve()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
DB_PATH = PROJECT_ROOT / "data" / "exoplanets.duckdb"
ART_DIR = PROJECT_ROOT / "artifacts"
DOCS_DIR = PROJECT_ROOT / "docs"

RAW_DIR.mkdir(parents=True, exist_ok=True)
ART_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

con = duckdb.connect(str(DB_PATH))

raw_csv = RAW_DIR / "pscomppars.csv"
if not raw_csv.exists():
    raise FileNotFoundError(f"No encuentro {raw_csv}. Necesitas el CSV de W01/W02.")

def sql_quote(s: str) -> str:
    return "'" + s.replace("'", "''") + "'"

con.execute(f'''
CREATE OR REPLACE VIEW raw_ps AS
SELECT * FROM read_csv_auto({sql_quote(str(raw_csv.resolve()))})
''')
````

````python
con.execute("DROP TABLE IF EXISTS silver_planet")
con.execute('''
CREATE TABLE silver_planet AS
SELECT
  pl_name, hostname, discoverymethod, disc_year,
  sy_snum, sy_pnum, sy_dist, ra, dec,
  pl_orbper, pl_rade, pl_bmasse, pl_eqt,
  st_teff, st_rad, st_mass
FROM raw_ps
WHERE pl_name IS NOT NULL
  AND hostname IS NOT NULL
  AND (disc_year IS NULL OR (disc_year BETWEEN 1980 AND 2026))
  AND (pl_rade  IS NULL OR (pl_rade  > 0 AND pl_rade  <= 30))
  AND (pl_bmasse IS NULL OR (pl_bmasse > 0))
''')

con.execute("DROP TABLE IF EXISTS dim_host_full")
con.execute('''
CREATE TABLE dim_host_full AS
SELECT
  hostname,
  MAX(sy_dist)  AS sy_dist,
  MAX(ra)       AS ra,
  MAX(dec)      AS dec,
  MAX(st_teff)  AS st_teff,
  MAX(st_rad)   AS st_rad,
  MAX(st_mass)  AS st_mass
FROM silver_planet
GROUP BY hostname
''')

con.execute("DROP TABLE IF EXISTS fact_planet")
con.execute('''
CREATE TABLE fact_planet AS
SELECT DISTINCT
  pl_name, hostname, discoverymethod, disc_year,
  pl_orbper, pl_rade, pl_bmasse, pl_eqt
FROM silver_planet
''')

con.sql("SELECT COUNT(*) AS n_fact, COUNT(DISTINCT pl_name) AS n_pl, COUNT(DISTINCT hostname) AS n_hosts FROM fact_planet").show()
````

````python
from datetime import datetime, timezone

# --- Asegurar conexión y directorios ---
try:
    con.execute("SELECT 1")
except Exception:
    con = duckdb.connect(str(DB_PATH))
ART_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

# --- 1. Consulta a analizar ---
q = """
SELECT 
    h.hostname,
    h.st_teff,
    COUNT(f.pl_name) AS n_planets,
    AVG(f.pl_rade) AS avg_radius
FROM fact_planet f
JOIN dim_host_full h ON f.hostname = h.hostname
GROUP BY h.hostname, h.st_teff
ORDER BY n_planets DESC
"""

# --- 2. EXPLAIN ANALYZE ---
explain_df = con.execute("EXPLAIN ANALYZE " + q).fetchdf()
explain_text = explain_df.to_string()

explain_file = ART_DIR / "w04a_explain_q.txt"
explain_file.write_text(explain_text, encoding="utf-8")
print("Plan guardado en:", explain_file)

# --- 3. Métricas ---
n_fact = con.execute("SELECT COUNT(*) FROM fact_planet").fetchone()[0]
n_dim  = con.execute("SELECT COUNT(*) FROM dim_host_full").fetchone()[0]

# --- 4. Reporte (usando + en vez de f-string multilinea, 100% seguro) ---
run_ts = datetime.now(timezone.utc).isoformat()

lineas = [
    "# W04A - Reporte de Performance",
    "",
    "**Timestamp:** " + run_ts,
    "",
    "## Tablas",
    "| Tabla | Filas |",
    "|-------|-------|",
    "| fact_planet | " + str(n_fact) + " |",
    "| dim_host_full | " + str(n_dim) + " |",
    "",
    "## Consulta analizada",
    "```sql",
    q.strip(),
    "```",
    "",
    "## Plan de ejecucion",
    "Guardado en: `" + str(explain_file.name) + "`",
    "",
    "---",
    "*Generado automaticamente*"
]

perf_report = "\n".join(lineas)

report_path = DOCS_DIR / "w04a_perf_report.md"
report_path.write_text(perf_report, encoding="utf-8")
print("Reporte guardado en:", report_path)
````
